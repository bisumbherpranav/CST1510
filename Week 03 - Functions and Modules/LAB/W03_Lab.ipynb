### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90): # function to determine status of percent
    # with the parameters percent and warning_at default at 90
    if percent >=100:       
        return "OVER LIMIT"     # returns OVER LIMIT if percent is greater than or equal to 100
    elif percent>=warning_at:   
        return "WARNING"     # returns WARNING if percent at 'warning_at' or more
    return "OK"         # if none applied, then returns OK
print(f"{status_of(95)}, {status_of(85)}, {status_of(85,80)}")
# call the function 3 times, with last one changing the default warning_at value to 80
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

````python
# D8
def total(*numbers):  # * means you can pass many numbers as you want
    """Returns the sum of all the numbers passed in total(numbers)"""
    total = 0
    for num in numbers:
        total+=num      
    return total   # loops through each number in numbers, add it to total and print the final total
print(total(1,2,3))
print(total(10,20,30,40))
print(total())      # no numbers passed -> total is 0

def configure(**settings):
    """Loops over each key and value in the dictionary and prints each one as key = value"""
    for key, value in settings.items():
        print(f"{key} = {value}")   
configure(host="10.0.0.1", port=8080)
# calls the configure function with the parameter with 2 keys (host and port) and 2 values ->
# 10.0.0.1 and 8080
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# C1
def percent_of(part,total):
    """Return part as a percentage of total"""
    return (part/total)*100     # returns percentage
def status_of(percent, warning_at=100):
    """Determines status of percentage, returns OVER LIMIT, WARNING, or OK"""
    if percent >=100:
        return "OVER LIMIT"     # returns OVER LIMIT if percent is 100 or more
    elif percent>=warning_at:
        return "WARNING"        # returns WARNING if percent is at warning or more
    return "OK"                 # if none applied, returns OK

# calls percent_of with part and total, returns the percentage
# then calls status_of with the percentage passed in the parameter, and returns status
print(status_of(percent_of(87,100)))
print(status_of(percent_of(120,100)))
print(status_of(percent_of(45,100)))
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit=100):     # *values, you can add as many values as you want
    """Counts how many of the values are over the limit=100"""
    count=0
    for value in values:       # loops through every value in the tuple
        if value > limit:      # check if value greater than limit
            count+=1           # add 1 to the counter for every over limit value
    return count
        
print(count_over(87,120,45,101))   
print(count_over(87,120,45,101, limit=50))   # changes the keyword argument limit to 50
print(count_over())   # no values passed -> returns 0
````
