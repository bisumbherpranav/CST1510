## Topic 4 · Parsing real text

_Walkthrough: `04 - Parsing Real Text`_


### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`


````python
# D14
raw_line = " Srv-01 , used=87 , total=100 "
line = (
    raw_line.strip()
)  # removes spaces from the beginning and the end, not the spaces inside string
hostname, rest = line.split(",", 1)  # split at the first comma
hostname = (
    hostname.strip().lower()
)  # removes spaces from the end and convert to lowercase
values = {}  # creates an empty dict
for item in rest.split(","):  # rest.split(",") creates a list for items in rest
    # items -> " used=87 " and " total=100"
    key, value = item.split("=")  # split at "=" -> used = key and 87 = value
    values[key.strip().lower()] = float(value.strip())
print({"hostname": hostname, "used": values["used"], "total": values["total"]})
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`


````python
# D15
def parse_line(raw_line):
    line = (
        raw_line.strip()
    )  # removes spaces from the beginning and the end, not the spaces inside string
    hostname, rest = line.split(",", 1)  # split at the first comma
    hostname = (
        hostname.strip().lower()
    )  # removes spaces from the end and convert to lowercase
    values = {}  # creates an empty dict
    for item in rest.split(","):  # rest.split(",") creates a list for items in rest
        # items -> " used=87 " and " total=100"
        key, value = item.split("=")  # split at "=" -> used = key and 87 = value
        values[key.strip().lower()] = float(value.strip())
    return {"hostname": hostname, "used": values["used"], "total": values["total"]}


parse_line(" Srv-02 , used=120 , total=100 ")
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**

```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```


````python
# C1
def status_of(used, total):     # define status_of takes used and total value 
    """Return OVER LIMIT, WARNING or OK for a used/total pair"""
    percent = (used / total) * 100      # calculates percentage
    if percent >= 100:
        return "OVER LIMIT"     # returns OVER LIMIT if percent 100 or more
    elif percent >= 90:
        return "WARNING"        # returns WARNING if percent 90 or more
    else:
        return "OK"             # returns OK if none applied


with open("messy_lines.txt") as f:      # opens "messy_lines.txt" file
    for line in f:                  # loop line by line
        record = parse_line(line)   # parse each line with parse_line function and store in record var
        print({record["hostname"], status_of(record["used"], record["total"])})
        # print hostname with its status - calls status_of with used and total from record
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**

```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}
Could not parse: srv-08,used=unknown,total=100
Could not parse: srv-09,note=a=b,used=1,total=2
```


````python
# C2
### C2. When parsing breaks
lines = [
    "srv-07,used=50,total=100",
    "srv-08,used=unknown,total=100",
    "srv-09,note=a=b,used=1,total=2",
]
for line in lines:      # loop through each line in lines
    try:
        record = parse_line(line) 
        # try to parse each line with parse_line function and store in record var
        print(
            {
                "hostname": record["hostname"],
                "used": record["used"],
                "total": record["total"],
            } # print hostname, used and total and put everything in a dict
        )
    except ValueError:  # if ValueError
        print(f"Could not parse: {line}")   # print "Could not parse: " with the line
````
